```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.model_selection import train_test_split, cross_val_score
from sklearn.neighbors import KNeighborsClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.neural_network import MLPClassifier
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
import seaborn as sns
import matplotlib.pyplot as plt
import joblib

# Load the dataset
data = pd.read_csv('Crop_recommendation.csv')

# Inspect the dataset
print("Step 1: Dataset Info")
print(data.info())
print("\nFirst 5 rows:")
print(data.head())
print("\nSummary statistics:")
print(data.describe())
print("\nUnique crop labels:", data['label'].unique())
```

```python
print("\nStep 2.1: Encoding Target Variable")
label_encoder = LabelEncoder()
data['label_encoded'] = label_encoder.fit_transform(data['label'])
label_mapping = dict(zip(label_encoder.classes_, label_encoder.transform(label_encoder.classes_)))
print("Label Mapping:", label_mapping)
```

```python
print("\nStep 2.2: Checking for Outliers")
numerical_cols = ['N', 'P', 'K', 'temperature', 'humidity', 'ph', 'rainfall']
def detect_outliers(df, column):
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    outliers = df[(df[column] < lower_bound) | (df[column] > upper_bound)][column]
    return len(outliers)
for col in numerical_cols:
    print(f"Outliers in {col}: {detect_outliers(data, col)}")
```

```python
print("\nStep 2.3: Splitting Features and Target")
X = data[numerical_cols]
y = data['label_encoded']
print("Features shape:", X.shape)
print("Target shape:", y.shape)
```

```python
print("\nStep 3.1: Feature Scaling")
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
X_scaled = pd.DataFrame(X_scaled, columns=numerical_cols)
print("Scaled features (first 5 rows):")
print(X_scaled.head())
```

```python
print("\nStep 4: Splitting Data into Training and Testing Sets")
X_train, X_test, y_train, y_test = train_test_split(X_scaled, y, test_size=0.2, random_state=42, stratify=y)
print("Training set shape:", X_train.shape, y_train.shape)
print("Testing set shape:", X_test.shape, y_test.shape)
```

```python
print("\nStep 5: KNN Model")
k_values = range(1, 21)
cv_scores_knn = [cross_val_score(KNeighborsClassifier(n_neighbors=k), X_train, y_train, cv=5, scoring='accuracy').mean() for k in k_values]
best_k = k_values[np.argmax(cv_scores_knn)]
print(f"Best k: {best_k} with CV Accuracy: {max(cv_scores_knn):.4f}")
knn = KNeighborsClassifier(n_neighbors=best_k)
knn.fit(X_train, y_train)
y_pred_knn = knn.predict(X_test)
accuracy_knn = accuracy_score(y_test, y_pred_knn)
print(f"KNN Test Accuracy: {accuracy_knn:.4f}")
print("KNN Classification Report:")
print(classification_report(y_test, y_pred_knn, target_names=label_encoder.classes_))
cm_knn = confusion_matrix(y_test, y_pred_knn)
```

```python
print("\nStep 5: KNN Model")
k_values = range(1, 21)
cv_scores_knn = [cross_val_score(KNeighborsClassifier(n_neighbors=k), X_train, y_train, cv=5, scoring='accuracy').mean() for k in k_values]
best_k = k_values[np.argmax(cv_scores_knn)]
print(f"Best k: {best_k} with CV Accuracy: {max(cv_scores_knn):.4f}")
knn = KNeighborsClassifier(n_neighbors=best_k)
knn.fit(X_train, y_train)
y_pred_knn = knn.predict(X_test)
accuracy_knn = accuracy_score(y_test, y_pred_knn)
print(f"KNN Test Accuracy: {accuracy_knn:.4f}")
print("KNN Classification Report:")
print(classification_report(y_test, y_pred_knn, target_names=label_encoder.classes_))
cm_knn = confusion_matrix(y_test, y_pred_knn)
print("KNN Confusion Matrix:")
print(cm_knn)
```

```python
print("\nStep 6: Decision Tree Model")
depth_values = range(3, 21)
cv_scores_dt = [cross_val_score(DecisionTreeClassifier(max_depth=depth, random_state=42), X_train, y_train, cv=5, scoring='accuracy').mean() for depth in depth_values]
best_depth = depth_values[np.argmax(cv_scores_dt)]
print(f"Best max_depth: {best_depth} with CV Accuracy: {max(cv_scores_dt):.4f}")
dt = DecisionTreeClassifier(max_depth=best_depth, random_state=42)
dt.fit(X_train, y_train)
y_pred_dt = dt.predict(X_test)
accuracy_dt = accuracy_score(y_test, y_pred_dt)
print(f"Decision Tree Test Accuracy: {accuracy_dt:.4f}")
print("Decision Tree Classification Report:")
print(classification_report(y_test, y_pred_dt, target_names=label_encoder.classes_))
cm_dt = confusion_matrix(y_test, y_pred_dt)
feature_importance_dt = pd.Series(dt.feature_importances_, index=numerical_cols).sort_values(ascending=False)
print("Decision Tree Feature Importance:\n", feature_importance_dt)
```

```python
print("\nStep 7: Logistic Regression Model")
c_values = [0.001, 0.01, 0.1, 1, 10, 100]
cv_scores_lr = [cross_val_score(LogisticRegression(C=c, max_iter=1000, random_state=42), X_train, y_train, cv=5, scoring='accuracy').mean() for c in c_values]
best_c = c_values[np.argmax(cv_scores_lr)]
print(f"Best C: {best_c} with CV Accuracy: {max(cv_scores_lr):.4f}")
lr = LogisticRegression(C=best_c, max_iter=1000, random_state=42)
lr.fit(X_train, y_train)
y_pred_lr = lr.predict(X_test)
accuracy_lr = accuracy_score(y_test, y_pred_lr)
print(f"Logistic Regression Test Accuracy: {accuracy_lr:.4f}")
print("Logistic Regression Classification Report:")
print(classification_report(y_test, y_pred_lr, target_names=label_encoder.classes_))
cm_lr = confusion_matrix(y_test, y_pred_lr)
feature_importance_lr = pd.Series(np.abs(lr.coef_).mean(axis=0), index=numerical_cols).sort_values(ascending=False)
print("Logistic Regression Feature Importance (Abs Coef):\n", feature_importance_lr)
```

```python
print("\nStep 8: Model Comparison with Bar Chart for Accuracy")
accuracies = {
    'KNN': accuracy_knn,
    'Decision Tree': accuracy_dt,
    'Logistic Regression': accuracy_lr
}
print("Model Accuracies:")
for model, acc in accuracies.items():
    print(f"{model}: {acc:.4f}")

# Bar chart for accuracies
plt.figure(figsize=(8, 6))
models = list(accuracies.keys())
acc_values = list(accuracies.values())
bars = plt.bar(models, acc_values, color=['blue', 'green', 'orange'])
plt.title('Model Accuracy Comparison', pad=15)
plt.xlabel('Models')
plt.ylabel('Test Accuracy')
plt.ylim(0.9, 1.0)  # Set y-axis limit to focus on high accuracies
# Add accuracy values on top of each bar
for bar in bars:
    height = bar.get_height()
    plt.text(bar.get_x() + bar.get_width()/2., height,
             f'{height:.4f}',
             ha='center', va='bottom')

plt.tight_layout()
plt.show()
```

```python
print("\nStep 9: Visualizing Feature Distributions (Histograms)")
plt.figure(figsize=(15, 10))
for i, col in enumerate(numerical_cols, 1):
    plt.subplot(3, 3, i)
    sns.histplot(data[col], az=30, kde=True)
    plt.title(f'Distribution of {col}')
    plt.xlabel(col)
    plt.ylabel('Count')
plt.tight_layout()
plt.show()
```

```python
print("\nStep 10: Visualizing Feature Importance")
plt.figure(figsize=(12, 5))
plt.subplot(1, 2, 1)
sns.barplot(x=feature_importance_dt.values, y=feature_importance_dt.index)
plt.title('Decision Tree Feature Importance')
plt.xlabel('Importance')
plt.subplot(1, 2, 2)
sns.barplot(x=feature_importance_lr.values, y=feature_importance_lr.index)
plt.title('Logistic Regression Feature Importance (Abs Coef)')
plt.xlabel('Importance')
plt.tight_layout()
plt.show()
```

```python
print("\nStep 11: Saving Models and Preprocessing Objects")
joblib.dump(knn, 'knn_crop_model.pkl')
joblib.dump(dt, 'decision_tree_crop_model.pkl')
joblib.dump(lr, 'logistic_regression_crop_model.pkl')
joblib.dump(scaler, 'scaler.pkl')
joblib.dump(label_encoder, 'label_encoder.pkl')
print("Models, scaler, and label encoder saved successfully.")
```

```python
print("\nStep 7: Multilayer Perceptron (MLP) Model")
# Define hyperparameter grid
hidden_sizes = [(100,), (50, 50), (100, 50)]
alphas = [0.0001, 0.001, 0.01]
learning_rates = [0.001, 0.01]
best_score = 0
best_params = {}

# Grid search with cross-validation
for hidden_size in hidden_sizes:
    for alpha in alphas:
        for learning_rate in learning_rates:
            mlp = MLPClassifier(hidden_layer_sizes=hidden_size, alpha=alpha, learning_rate_init=learning_rate,
                              max_iter=500, random_state=42, early_stopping=True)
            scores = cross_val_score(mlp, X_train, y_train, cv=5, scoring='accuracy')
            mean_score = scores.mean()
            if mean_score > best_score:
                best_score = mean_score
                best_params = {'hidden_layer_sizes': hidden_size, 'alpha': alpha, 'learning_rate_init': learning_rate}
print(f"Best parameters: {best_params} with CV Accuracy: {best_score:.4f}")

# Train final model
mlp = MLPClassifier(hidden_layer_sizes=best_params['hidden_layer_sizes'], alpha=best_params['alpha'],
                    learning_rate_init=best_params['learning_rate_init'], max_iter=500, random_state=42,
                    early_stopping=True)
mlp.fit(X_train, y_train)
y_pred_mlp = mlp.predict(X_test)
accuracy_mlp = accuracy_score(y_test, y_pred_mlp)
print(f"MLP Test Accuracy: {accuracy_mlp:.4f}")
print("MLP Classification Report:")
print(classification_report(y_test, y_pred_mlp, target_names=label_encoder.classes_))
cm_mlp = confusion_matrix(y_test, y_pred_mlp)

# Feature importance approximation (using input weights)
feature_importance_mlp = pd.Series(np.abs(mlp.coefs_[0]).mean(axis=1), index=numerical_cols).sort_values(ascending=False)
print("MLP Feature Importance (Abs Input Weights):\n", feature_importance_mlp)
```

```python
print("\nStep 8: Model Comparison with Bar Chart for Accuracy")
accuracies = {
    'Decision Tree': accuracy_dt,
    'Logistic Regression': accuracy_lr,
    'MLP': accuracy_mlp
}
print("Model Accuracies:")
for model, acc in accuracies.items():
    print(f"{model}: {acc:.4f}")

# Bar chart for accuracies
plt.figure(figsize=(8, 6))
models = list(accuracies.keys())
acc_values = list(accuracies.values())
bars = plt.bar(models, acc_values, color=['green', 'orange', 'purple'])
plt.title('Model Accuracy Comparison', pad=15)
plt.xlabel('Models')
plt.ylabel('Test Accuracy')
plt.ylim(0.9, 1.0)  # Focus on high accuracies
# Add accuracy values on top of each bar
for bar in bars:
    height = bar.get_height()
    plt.text(bar.get_x() + bar.get_width()/2., height,
             f'{height:.4f}',
             ha='center', va='bottom')
plt.tight_layout()
plt.show()
```

```python
print("\nStep 10: Visualizing Feature Importance")
plt.figure(figsize=(18, 5))  # Increased width to accommodate three subplots
plt.subplot(1, 3, 1)
sns.barplot(x=feature_importance_dt.values, y=feature_importance_dt.index)
plt.title('Decision Tree Feature Importance')
plt.xlabel('Importance')

plt.subplot(1, 3, 2)
sns.barplot(x=feature_importance_lr.values, y=feature_importance_lr.index)
plt.title('Logistic Regression Feature Importance (Abs Coef)')
plt.xlabel('Importance')

plt.subplot(1, 3, 3)
sns.barplot(x=feature_importance_mlp.values, y=feature_importance_mlp.index)
plt.title('MLP Feature Importance (Abs Input Weights)')
plt.xlabel('Importance')

plt.tight_layout()
plt.show()
```

```python
print("\nStep 12: Crop Prediction")
def predict_crop():
    # Get input from user
    print("Enter the following feature values for prediction:")
    n = float(input("N (Nitrogen, 0-140): "))
    p = float(input("P (Phosphorus, 5-145): "))
    k = float(input("K (Potassium, 5-205): "))
    temperature = float(input("Temperature (14-44°C): "))
    humidity = float(input("Humidity (14-100%): "))
    ph = float(input("pH (3-10): "))
    rainfall = float(input("Rainfall (20-298 mm): "))

    # Create input array
    input_data = np.array([[n, p, k, temperature, humidity, ph, rainfall]])

    # Scale the input data
    input_scaled = scaler.transform(input_data)

    # Predict using each model
    pred_dt = dt.predict(input_scaled)[0]
    pred_lr = lr.predict(input_scaled)[0]
    pred_mlp = mlp.predict(input_scaled)[0]

    # Decode predictions to crop names
    crop_dt = label_encoder.inverse_transform([pred_dt])[0]
    crop_lr = label_encoder.inverse_transform([pred_lr])[0]
    crop_mlp = label_encoder.inverse_transform([pred_mlp])[0]

    # Print results
    print("\nPredicted Crop:")
    print(f"Decision Tree: {crop_dt}")
    print(f"Logistic Regression: {crop_lr}")
    print(f"MLP: {crop_mlp}")

# Run the prediction function
predict_crop()
```
